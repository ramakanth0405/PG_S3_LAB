```python
# List to keep track of visited nodes
visited = []
# Queue to store nodes that need to be explored next
queue = []

def bfs(visited, graph, node):
    # Start by visiting the root node and adding it to the queue
    visited.append(node)
    queue.append(node)

    while queue:
        # Remove the first element from the queue (FIFO)
        m = queue.pop(0)
        print(m, end=" ")

        # Check all the neighbors of the current node
        for neighbour in graph[m]:
            if neighbour not in visited:
                visited.append(neighbour)
                queue.append(neighbour)
# Represent the graph using an adjacency list (a dictionary in Python)
graph = {
    'A': ['B', 'C'],
    'B': ['D', 'E'],
    'C': ['F'],
    'D': [],
    'E': ['F'],
    'F': []
}

bfs_in = input("Enter the starting node: ")
print("Breadth-First Search traversal path:")
bfs(visited, graph, bfs_in.upper())
```
