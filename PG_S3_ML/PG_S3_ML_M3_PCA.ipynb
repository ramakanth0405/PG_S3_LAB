### Using StandardScaler from scikit-learn to standardize features

```python
from sklearn.preprocessing import StandardScaler
scaler = StandardScaler()
```

```python
scaler.fit(df_1)
scaled_data = scaler.transform(df_1)
scaled_data
```

## Principal Component Analysis (PCA) to reduce dimensionality

### Using PCA to shrink the features to 1

```python
from sklearn.decomposition import PCA
pca_1D = PCA(n_components=1)
pca_1D
```

```python
data_pca_1D = pca_1D.fit_transform(scaled_data)
data_pca_1D
```

```python
import matplotlib.pyplot as plt
import numpy as np
import seaborn as sns
sns.scatterplot(
    x=data_pca_1D[:, 0], 
    y=np.zeros_like(data_pca_1D[:, 0]), 
    hue=cancer_data['target']
)
plt.show()
```

### Using PCA to shrink the features to 2 

```python
from sklearn.decomposition import PCA
pca_1 = PCA(n_components=2)
pca_1
```

```python
data_pca_1 = pca_1.fit_transform(scaled_data)
data_pca_1
```

```python
pca_1.explained_variance_
```

### Using PCA to shrink the features to 3

```python
pca_2 = PCA(n_components=3)
pca_2
```

```python
data_pca_2 = pca_2.fit_transform(scaled_data)
data_pca_2
```

```python
pca_2.explained_variance_ratio_
```

```python
import matplotlib.pyplot as plt
sns.scatterplot(x = data_pca_1[:,0], y=data_pca_1[:,1], data = cancer_data, hue='target')
```

```python
import matplotlib.pyplot as plt
from mpl_toolkits.mplot3d import Axes3D

fig = plt.figure(figsize=(8, 6))
ax = fig.add_subplot(111, projection='3d')

ax.scatter(
    data_pca_2[:, 0],
    data_pca_2[:, 1],
    data_pca_2[:, 2],
    c=cancer_data.target,
    cmap='viridis',
    s=50
)

ax.set_title('PCA with 3 Components')
ax.set_xlabel('Principal Component 1')
ax.set_ylabel('Principal Component 2')
ax.set_zlabel('Principal Component 3')
plt.show()
```

## Model Training

```python
from sklearn.model_selection import train_test_split
from sklearn.linear_model import LogisticRegression
from sklearn.metrics import accuracy_score, classification_report
from sklearn.metrics import confusion_matrix
```

### Assining Target variable

```python
Y = cancer_data.target
```

### Training Model with 2 components

```python
X_train_2, X_test_2, y_train_2, y_test_2 = train_test_split(
    data_pca_1, Y, test_size=0.2, random_state=42, stratify=Y
)

logreg_2 = LogisticRegression(max_iter=1000, random_state=42)
logreg_2.fit(X_train_2, y_train_2)
y_pred_2 = logreg_2.predict(X_test_2)

print('Logistic Regression with PCA 2 components')
print('Accuracy:', accuracy_score(y_test_2, y_pred_2))
print(classification_report(y_test_2, y_pred_2))
print('-' * 60)
```

### Confusion matrix for Model Trained with 2-component PCA

```python
cm_2 = confusion_matrix(y_test_2, y_pred_2)

plt.figure(figsize=(5, 4))
sns.heatmap(
    cm_2,
    annot=True,
    cmap='Blues',
    xticklabels=['No Cancer', 'Cancer'],
    yticklabels=['No Cancer', 'Cancer']
)
plt.title('Confusion Matrix - Logistic Regression with PCA 2 components')
plt.xlabel('Predicted')
plt.ylabel('Actual')
plt.show()
```

### Training Model with 3 components

```python
X_train_3, X_test_3, y_train_3, y_test_3 = train_test_split(
    data_pca_2, Y, test_size=0.2, random_state=42, stratify=Y
)

logreg_3 = LogisticRegression(max_iter=1000, random_state=42)
logreg_3.fit(X_train_3, y_train_3)
y_pred_3 = logreg_3.predict(X_test_3)

print('Logistic Regression with PCA 3 components')
print('Accuracy:', accuracy_score(y_test_3, y_pred_3))
print(classification_report(y_test_3, y_pred_3))
```

### Confusion matrix for Model Trained with 3-component PCA

```python
cm_3 = confusion_matrix(y_test_3, y_pred_3)

plt.figure(figsize=(5, 4))
sns.heatmap(
    cm_3,
    annot=True,
    fmt='d',
    cmap='Blues',
    cbar=False,
    xticklabels=['No Cancer', 'Cancer'],
    yticklabels=['No Cancer', 'Cancer']
)
plt.title('Confusion Matrix - Logistic Regression with PCA 3 components')
plt.xlabel('Predicted')
plt.ylabel('Actual')
plt.show()
```
