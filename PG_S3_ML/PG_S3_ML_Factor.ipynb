```python
import pandas as pd
from factor_analyzer import FactorAnalyzer
```

```python
# Load dataset
data = pd.read_csv("/Users/ramakanthvprabhu/Desktop/PG_S3/Datasets/student_data.csv")
```

```python
# Create Factor Analyzer
fa = FactorAnalyzer(n_factors=3, rotation='varimax')
 
# Fit the model
fa.fit_transform(data)
 
# Get factor loadings
loadings = fa.loadings_
 
print("Factor Loadings:")
print(loadings)
 
# Get eigenvalues
eigenvalues, vectors = fa.get_eigenvalues()
 
print("Eigenvalues:")
print(eigenvalues)
```
