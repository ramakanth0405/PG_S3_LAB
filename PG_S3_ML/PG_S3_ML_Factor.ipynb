```python
import pandas as pd
from factor_analyzer import FactorAnalyzer
import matplotlib.pyplot as plt
import seaborn as sns
```

```python
# Load dataset
data = pd.read_csv("/Users/ramakanthvprabhu/Desktop/PG_S3/Datasets/student_data.csv")
```

```python
# Create Factor Analyzer
fa = FactorAnalyzer(n_factors=3, rotation='varimax')
 
# Fit the model
fa.fit_transform(data)
 
# Get factor loadings
loadings = fa.loadings_
 
print("Factor Loadings:")
print(loadings)
 
# Get eigenvalues
eigenvalues, vectors = fa.get_eigenvalues()
 
print("Eigenvalues:")
print(eigenvalues)
```

```python
# 1. Scree Plot (for Eigenvalues)
plt.figure(figsize=(8, 4))
plt.plot(range(1, len(eigenvalues) + 1), eigenvalues, marker='o', color='b', linestyle='-')
plt.title('Scree Plot')
plt.xlabel('Factors')
plt.ylabel('Eigenvalues')
plt.axhline(y=1, color='r', linestyle='--', label='Kaiser Rule (Eigenvalue = 1)')
plt.grid(True)
plt.legend()
plt.show()

# 2. Factor Loadings Heatmap
# Creates a structured DataFrame representing the loadings
loadings_df = pd.DataFrame(
    loadings, 
    index=data.columns, 
    columns=[f'Factor {i+1}' for i in range(loadings.shape[1])]
)

plt.figure(figsize=(8, 6))
sns.heatmap(loadings_df, annot=True, cmap='coolwarm', vmin=-1, vmax=1, fmt=".2f")
plt.title('Factor Loadings Heatmap')
plt.ylabel('Variables')
plt.xlabel('Factors')
plt.show()
```
